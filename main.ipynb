**Step 2: Model**

Linear regression, four required DNNs, additional DNN-32-16-output. Use the same approximately 70%/15%/15% split and 30 numeric inputs for every model. Random seed 67. The separate `test.csv` remains reserved for future predictions. This is a first comparison with a fixed 500-epoch budget, not a completed search for the best model.

```python
from copy import deepcopy
from pathlib import Path

import numpy as np
import pandas as pd
from IPython.display import display
from sklearn.compose import ColumnTransformer
from sklearn.linear_model import SGDRegressor
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.model_selection import train_test_split
from sklearn.neural_network import MLPRegressor
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from threadpoolctl import threadpool_limits

SEED = 123
EPOCHS = 500
TARGET = 'TARGET_deathRate'

# Use the same cleanup already used in the notebook.
raw_train = pd.read_csv('train.csv').drop(columns=['PctSomeCol18_24'])
for column in ['PctEmployed16_Over', 'PctPrivateCoverageAlone']:
    raw_train[column] = raw_train[column].fillna(raw_train[column].median())

# Geography is excluded because no test county names occur in training data.
# binnedInc is kept and encoded below.
X = raw_train.drop(columns=[TARGET, 'id', 'Geography'])
y = raw_train[TARGET]

X_train, X_holdout, y_train, y_holdout = train_test_split(
    X, y, test_size=0.30, random_state=SEED
)
X_val, X_test, y_val, y_test = train_test_split(
    X_holdout, y_holdout, test_size=0.50, random_state=SEED
)
```

```python
# Scale numeric columns and one-hot encode binnedInc.
numeric_columns = X_train.select_dtypes(include='number').columns
text_columns = X_train.select_dtypes(exclude='number').columns
preprocessor = ColumnTransformer([
    ('numeric', StandardScaler(), numeric_columns),
    ('text', OneHotEncoder(handle_unknown='ignore', sparse_output=False), text_columns),
])
X_train_scaled = preprocessor.fit_transform(X_train)
X_val_scaled = preprocessor.transform(X_val)
X_test_scaled = preprocessor.transform(X_test)

target_scaler = StandardScaler()
y_train_scaled = target_scaler.fit_transform(y_train.to_numpy().reshape(-1, 1)).ravel()


def predict_death_rate(model, features):
    return target_scaler.inverse_transform(
        model.predict(features).reshape(-1, 1)
    ).ravel()
```

```python
architectures = {
    'DNN-8-output': (8,),
    'DNN-16-8-output': (16, 8),
    'DNN-16-8-4-output': (16, 8, 4),
    'DNN-30-16-8-4-output': (30, 16, 8, 4),
    'DNN-32-16-output (additional)': (32, 16),
}


def make_models():
    models = {
        'Linear regression': SGDRegressor(
            loss='squared_error', penalty=None, random_state=SEED
        ),
    }
    for name, layers in architectures.items():
        models[name] = MLPRegressor(
            hidden_layer_sizes=layers, activation='relu', solver='sgd',
            alpha=0.0, random_state=SEED
        )
    return models
```

**Train and select checkpoints using validation MSE**

`partial_fit` performs one epoch while retaining optimizer state. Train each architecture for 500 epochs and save its lowest-validation-MSE checkpoint. The local test targets are not used during training or checkpoint selection. Keep the full training and validation MSE histories for the later Step 5 figure.

```python
best_models = {}
best_epochs = {}
best_validation_mse = {}
history_rows = []

# Small networks are faster with one numerical-library thread on this dataset.
with threadpool_limits(limits=1):
    for name, model in make_models().items():
        lowest_val_mse = np.inf
        for epoch in range(1, EPOCHS + 1):
            model.partial_fit(X_train_scaled, y_train_scaled)
            train_predictions = predict_death_rate(model, X_train_scaled)
            val_predictions = predict_death_rate(model, X_val_scaled)
            train_mse = mean_squared_error(y_train, train_predictions)
            val_mse = mean_squared_error(y_val, val_predictions)
            history_rows.append({
                "Model": name,
                "Epoch": epoch,
                "Train MSE": train_mse,
                "Validation MSE": val_mse,
            })
            if val_mse < lowest_val_mse:
                lowest_val_mse = val_mse
                best_models[name] = deepcopy(model)
                best_epochs[name] = epoch
                best_validation_mse[name] = val_mse

        print(f"{name}: selected epoch {best_epochs[name]}, "
              f"validation MSE = {lowest_val_mse:.3f}")

history = pd.DataFrame(history_rows)
# Select the architecture before inspecting its local test performance.
selected_model_name = min(best_validation_mse, key=best_validation_mse.get)
selected_model = best_models[selected_model_name]
print(f"\nSelected by validation MSE: {selected_model_name}")
```

```python
comparison_rows = []
for name, model in best_models.items():
    row = {"Model": name, "Best epoch": best_epochs[name]}
    for split, features, labels in [
        ("Train", X_train_scaled, y_train),
        ("Validation", X_val_scaled, y_val),
        ("Test", X_test_scaled, y_test),
    ]:
        predictions = predict_death_rate(model, features)
        row[f"{split} R-squared"] = r2_score(labels, predictions)
        row[f"{split} MSE"] = mean_squared_error(labels, predictions)
    comparison_rows.append(row)

comparison = pd.DataFrame(comparison_rows)
comparison["Train-validation R-squared gap"] = (
    comparison["Train R-squared"] - comparison["Validation R-squared"]
)

# The first table matches the requested Step 2 performance table.
display(comparison[["Model", "Test R-squared"]].round(4))
# These diagnostics support the bias/variance discussion and later MSE report.
display(comparison.round(4))
```

**9) Initial bias and variance observations**

The following summary uses the measured training and validation results. A training/validation gap is evidence consistent with overfitting, but a single split cannot measure variance across datasets. Low training performance can also reflect incomplete optimization, so do not label a model as underfitting solely because its test score is low. Architectures selected at epoch 500 may still benefit from additional training; this initial run does not establish convergence.

```python
validation_order = comparison.sort_values("Validation MSE")
best_row = validation_order.iloc[0]
worst_row = validation_order.iloc[-1]
largest_gap = comparison.loc[comparison["Train-validation R-squared gap"].idxmax()]
best_test = comparison.loc[comparison["Test R-squared"].idxmax()]
worst_test = comparison.loc[comparison["Test R-squared"].idxmin()]

print(f"Best by validation: {best_row['Model']} "
      f"(validation R-squared {best_row['Validation R-squared']:.4f}; "
      f"local test R-squared {best_row['Test R-squared']:.4f}).")
print(f"Lowest validation score: {worst_row['Model']} "
      f"(R-squared {worst_row['Validation R-squared']:.4f}).")
print(f"Highest local test score: {best_test['Model']} "
      f"({best_test['Test R-squared']:.4f}); "
      f"lowest: {worst_test['Model']} ({worst_test['Test R-squared']:.4f}).")
print(f"Largest train-validation gap: {largest_gap['Model']} "
      f"({largest_gap['Train-validation R-squared gap']:.4f}); "
      "this is consistent with overfitting on this split.")
print("The local test ranking is descriptive; model selection used validation MSE.")
print("Capacity, initialization, the finite training budget, and the numeric-only "
      "feature set may explain differences. More layers do not guarantee better performance.")
```

**Step 3: Objective**

MSE is the required loss function. This experiment also tries Huber loss with the linear SGD model. Huber loss uses squared error for small errors and reduces the effect of very large errors.

```python
# Compare the required MSE objective with a different loss function.
mse_model = SGDRegressor(
    loss='squared_error', penalty=None, random_state=SEED
)
huber_model = SGDRegressor(
    loss='huber', penalty=None, random_state=SEED
)

loss_results = []
for name, model in [('MSE', mse_model), ('Huber', huber_model)]:
    model.fit(X_train_scaled, y_train_scaled)
    predictions = predict_death_rate(model, X_test_scaled)
    loss_results.append({
        'Loss function': name,
        'Test R-squared': r2_score(y_test, predictions),
        'Test MSE': mean_squared_error(y_test, predictions),
    })

loss_results = pd.DataFrame(loss_results)
loss_results.round(4)
```

**10. Different loss function**

Huber loss was tested in addition to MSE. It is less sensitive to large prediction errors because its penalty becomes linear for larger residuals. In this run, MSE performed better: its test MSE was 308.5211 and test R-squared was 0.5918, compared with Huber's test MSE of 311.8329 and R-squared of 0.5874. The smaller Huber penalty on large errors did not improve this dataset's test performance.

```python
# Train the validation-selected DNN on all cleaned training rows,
# then write predictions in Kaggle's required format.
kaggle_test = pd.read_csv('test.csv').drop(columns=['PctSomeCol18_24'])
for column in ['PctEmployed16_Over', 'PctPrivateCoverageAlone']:
    kaggle_test[column] = kaggle_test[column].fillna(raw_train[column].median())

full_X = raw_train.drop(columns=[TARGET, 'id', 'Geography'])
full_y = raw_train[TARGET]
kaggle_X = kaggle_test.drop(columns=['id', 'Geography'])

submission_preprocessor = ColumnTransformer([
    ('numeric', StandardScaler(), numeric_columns),
    ('text', OneHotEncoder(handle_unknown='ignore', sparse_output=False), text_columns),
])
full_X = submission_preprocessor.fit_transform(full_X)
kaggle_X = submission_preprocessor.transform(kaggle_X)

submission_target_scaler = StandardScaler()
full_y = submission_target_scaler.fit_transform(full_y.to_numpy().reshape(-1, 1)).ravel()
if selected_model_name == 'Linear regression':
    submission_model = SGDRegressor(
        loss='squared_error', penalty=None, random_state=SEED
    )
else:
    selected_layers = architectures[selected_model_name]
    submission_model = MLPRegressor(
        hidden_layer_sizes=selected_layers, activation='relu', solver='sgd',
        alpha=0.0, random_state=SEED,
        max_iter=best_epochs[selected_model_name]
    )

submission_model.fit(full_X, full_y)
predictions = submission_target_scaler.inverse_transform(
    submission_model.predict(kaggle_X).reshape(-1, 1)
).ravel()

submission = pd.DataFrame({
    'id': kaggle_test['id'],
    'TARGET_deathRate': predictions,
})
submission.to_csv('submission.csv', index=False)
submission.head()
```
